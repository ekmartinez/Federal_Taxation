**Accelerating Deductions**

Mercedes, a calendar-year taxpayer, uses the cash method of accounting for her small business. On December 28, she receives a 10,000 bill from her accountant for consulting services related to her small business. She can avoid late payment charges by paying the 10,000 bill before January 10 of next year. Let’s assume that Mercedes’s marginal tax rate is 32 percent this year and next and that she can earn an after-tax rate of return of 10 percent on her investments. When should she pay the 10,000 bill—this year or next?

**To solve this problem we need to get the After-tax cost of accounting services of each option.**

1. **Option 1** - Pay the bill this year.
2. **Option 2** - Pay the bill next year.

```python
bill = 10000
pmt = 0
discount_rate = .10
marginal_tax_rate = .32

tax_savings = RPNEngine(f"{bill} {marginal_tax_rate} *").safe_evaluate().value
print(f"The tax savings via business deduction is: ${tax_savings:,.2f}")

# Pay the bill this year
nper = 0
pv_of_tax_savings_cy= RPNEngine(f"{discount_rate} {nper} {pmt} {tax_savings} pv").safe_evaluate().value
print(f"The present value of option 1 is: ${-pv_of_tax_savings_cy:,.2f}")

# Pay the bill next year
nper = 1
pv_of_tax_savings_ny = RPNEngine(f"{discount_rate} {nper} {pmt} {tax_savings} pv").safe_evaluate().value
print(f"The present value of option 2 is: ${-pv_of_tax_savings_ny:,.2f}")

option_1 = RPNEngine(f"{bill} {-pv_of_tax_savings_cy} -").safe_evaluate().value
option_2 = RPNEngine(f"{bill} {-pv_of_tax_savings_ny} -").safe_evaluate().value

# After tax cost of accounting services
print(f"The after-tax cost of the bill if payed this year: ${option_1:,.2f}")
print(f"The after-tax cost of the bill if payed next year: ${option_2:,.2f}")
print(f"""\nSince Mercedes will have a lower after-tax cost paying this year,
she choses option 1""")
```

**Deferring Income**
 
In early December, Bill decides he would like to sell 100,000 of his Dell Inc. stock, which 10 years ago cost 20,000. Assume Bill’s tax rate on the 80,000 gain will be 15 percent and his typical after-tax rate of return on investments is 7 percent. What effect would deferring the sale to January have on Bill’s after-tax cash flow on the sale?

```python
discount_rate = .07
pmt = 0
sales_price = 100000
cost_of_stock = 20000
marginal_tax_rate = .15

gain = RPNEngine(f"{sales_price} {cost_of_stock} -").safe_evaluate().value
print(f"The gain on the sale of stock is: ${gain:,.0f}")
tax_on_gain = RPNEngine(f"{gain} {marginal_tax_rate} *").safe_evaluate().value
print(f"The tax cost on the gain is: ${tax_on_gain:,.0f}")

# Option 1 - Sell in December of current year
nper = 0
pv_of_tax_cost_cy = RPNEngine(f"{discount_rate} {nper} {pmt} {tax_on_gain} pv").safe_evaluate().value
print(f"\nThe present value of the tax cost if sold in current year: ${-pv_of_tax_cost_cy:,.0f}")

# Option 2 - Sell in January of next year
nper = 1
pv_of_tax_cost_ny = RPNEngine(f"{discount_rate} {nper} {pmt} {tax_on_gain} pv").safe_evaluate().value
print(f"The present value of the tax cost if sold next year: ${-pv_of_tax_cost_ny:,.0f}")

option_1 = RPNEngine(f"{sales_price} {-pv_of_tax_cost_cy} -").safe_evaluate().value
print(f"\nThe after-tax cost if sold in the current year is: ${option_1:,.0f}")
option_2 = RPNEngine(f"{sales_price} {-pv_of_tax_cost_ny} -").safe_evaluate().value
print(f"The after-tax cost if sold next year is: ${option_2:,.0f}")
print(f"""\nSince the After-tax cash flow from the sale is greater in option 2 (${option_2:,.0f}) 
rather than in option 1 (${option_1:,.0f}), Bill decides to make the transaction next year.""")
```

**Deductions When Tax Rates Are Changing**

Having decided she needs new equipment for her business, Mercedes is now considering whether to make the purchase and claim a corresponding 10,000 deduction at year-end or next year. Mercedes anticipates that, with the new machinery, her business income will rise such that her marginal rate will increase from 24 percent this year to 32 percent next year. Assuming her after-tax rate of return is 8 percent, what should Mercedes do?

```python
discount_rate = .08
deduction = 10000
pmt = 0

# Option 1 - Make the deduction this year
nper = 0
marginal_tax_rate = .24
tax_savings = RPNEngine(f"{deduction} {marginal_tax_rate} *").safe_evaluate().value
pv_of_savings_cy = RPNEngine(f"{discount_rate} {nper} {pmt} {tax_savings} pv").safe_evaluate().value
after_tax_cost_cy = RPNEngine(f"{deduction} {pv_of_savings_cy} +").safe_evaluate().value

print("Option 1")
print(f"The tax savings if the deduction is taken in the current year is: ${tax_savings:,.0f}")
print(f"The present value of the deduction if taken in the current year is: ${-pv_of_savings_cy:,.0f}")
print(f"The after-tax cost of the equipment if bought this year: ${after_tax_cost_cy:,.0f}")

# Option 2 - Make the deduction next year
nper = 1
marginal_tax_rate = .32
tax_savings = RPNEngine(f"{deduction} {marginal_tax_rate} *").safe_evaluate().value
pv_of_savings_ny = RPNEngine(f"{discount_rate} {nper} {pmt} {tax_savings} pv").safe_evaluate().value
after_tax_cost_ny = RPNEngine(f"{deduction} {pv_of_savings_ny} +").safe_evaluate().value
print("\nOption 2")
print(f"The tax savings if the deduction is taken next year is: ${tax_savings:,.0f}")
print(f"The present value of the deduction if taken next year is: ${-pv_of_savings_ny:,.0f}")
print(f"The after-tax cost of the equipment if bought next year: ${after_tax_cost_ny:,.0f}")

print(f"""\nSince the after-tax cost of the equipment is lower next year (${after_tax_cost_ny:,.0f})
rather that in the current year (${after_tax_cost_cy:,.0f}), Mercedes decides to wait
until next year the make the purchase.""")
```

**Income Shifting - Businesses**

Mercedes is considering organizing her interior design business as a C corporation. She projects 200,000 of business profit next year. Excluding this profit, Bill and Mercedes expect 70,000 of taxable income next year. If Mercedes would like to minimize her current-year tax liability, should she incorporate her business? 

```python
status = "MFJ"
corporate_tax_rate = .21
busines_profit = 200000
taxable_income = 70000

# Scenario 1 - No Incorporation
print("Scenario 1 - No Incorporation")
total_income = RPNEngine(f"{busines_profit} {taxable_income} +").safe_evaluate().value
print(f"Total income {total_income}")
ordinary_tax_liability = OrdinaryTax(status, total_income).compute_tax()
print(f"Tax liability: ${ordinary_tax_liability:,.0f}")

# Scenario 2 - With Incorporation
print("\nScenario 2 - Incorporation")
corporate_tax = RPNEngine(f"{busines_profit} {corporate_tax_rate} *").safe_evaluate().value
ordinary_tax_liability = OrdinaryTax(status, taxable_income).compute_tax()
total_tax = RPNEngine(f"{corporate_tax} {ordinary_tax_liability} +").safe_evaluate().value
print(f"Corporate tax liability: ${corporate_tax:,.0f}")
print(f"Ordinary tax liability: ${ordinary_tax_liability:,.0f}")
print(f"Total tax: ${total_tax:,.0f}")
```
