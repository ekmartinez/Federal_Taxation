Bill is given the choice of receiving a 1,000 nontaxable gift today or a 1,000 nontaxable gift in one year. Which would Bill prefer? Assume Bill could invest 1,000 today and earn an 8 percent return after taxes in one year. If he receives the gift today, how much would the 1,000 be worth in one year?

```python
rate = .08
nper = 1
pmt = 0
pv = -1000

rpn_fv = RPNEngine(f"{rate} {nper} {pmt} {pv} fv")
fv = rpn_fv.safe_evaluate().value

print(f"His investment of $1,000 will be worth ${fv:,.0f} in one year.")
```

Mercedes, a calendar-year taxpayer, uses the cash method of accounting for her small business. On December 28, she receives a 10,000 bill from her accountant for consulting services related to her small business. She can avoid late payment charges by paying the 10,000 bill before January 10 of next year. Let’s assume that Mercedes’s marginal tax rate is 32 percent this year and next and that she can earn an after-tax rate of return of 10 percent on her investments. When should she pay the 10,000 bill—this year or next?

**To solve this problem we need to get the After-tax cost of accounting services of each option.**

1. **Option 1** - Pay the bill this year.
2. **Option 2** - Pay the bill next year.

```python
bill = 10000
pmt = 0
discount_rate = .10
marginal_tax_rate = .32

tax_savings = RPNEngine(f"{bill} {marginal_tax_rate} *").safe_evaluate().value
print(f"The tax savings via business deduction is: ${tax_savings:,.2f}")

# Pay the bill this year
nper = 0
pv_of_tax_savings_cy= RPNEngine(f"{discount_rate} {nper} {pmt} {tax_savings} pv").safe_evaluate().value
print(f"The present value of option 1 is: ${-pv_of_tax_savings_cy:,.2f}")

# Pay the bill next year
nper = 1
pv_of_tax_savings_ny = RPNEngine(f"{discount_rate} {nper} {pmt} {tax_savings} pv").safe_evaluate().value
print(f"The present value of option 2 is: ${-pv_of_tax_savings_ny:,.2f}")

option_1 = RPNEngine(f"{bill} {-pv_of_tax_savings_cy} -").safe_evaluate().value
option_2 = RPNEngine(f"{bill} {-pv_of_tax_savings_ny} -").safe_evaluate().value

# After tax cost of accounting services
print(f"The after-tax cost of the bill if payed this year: ${option_1:,.2f}")
print(f"The after-tax cost of the bill if payed next year: ${option_2:,.2f}")
print(f"""\nSince Mercedes will have a lower after-tax cost paying this year,
she choses option 1""")
```
