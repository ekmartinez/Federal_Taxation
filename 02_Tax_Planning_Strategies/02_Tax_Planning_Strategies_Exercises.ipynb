```python
# This is a Deduction Timing Strategy problem when tax rates are constant, we already know that in this
# kind of situations, the tax payer is better off accelerating deductions into earlier years to maximize
# the present value of tax savings from deductions, but for illustration purposes we will provide support 
# to our conclusion.

category = "expense"
before_tax_amount = 20000
rate_of_return = .12

scenarios = {
    "Scenario_1": [0, .37], # This year
    "Scenario_2": [1, .37]  # Next year
}

timming_strategy(category, before_tax_amount, 0, rate_of_return, scenarios)

print(f"""Since Isabel prefers to pay $12,600 this year, rather 
than $13,393 next year, she chooses this year.""")
```

Moana is a single taxpayer who operates a sole proprietorship. She expects her taxable income next year to be 250,000, of which 200,000 is attributed to her sole proprietorship. Moana is contemplating incorporating her sole proprietorship. Using the 2026 single individual tax brackets, the 2026 corporate tax rate and ignoring any Social Security, Medicare, or self-employment tax issues find out:

- a) How much current tax this strategy could save Moana. 
- b) How much income should be left in the corporation?

```python
status = "SNG"
corporate_tax_rate = .21
total_income = 250000
business_income = 200000
personal_income = RPNEngine(f"{total_income} {business_income} -").safe_evaluate().value

# Current strategy Tax
current_strategy_tax =  OrdinaryTax(status, total_income).compute_tax()

# Corporation strategy tax
personal_tax = OrdinaryTax(status, personal_income).compute_tax()
corporate_tax = RPNEngine(f"{business_income} {corporate_tax_rate} *").safe_evaluate().value
corporation_strategy_tax = RPNEngine(f"{personal_tax} {corporate_tax} +").safe_evaluate().value

# Current Tax Savings
savings = RPNEngine(f"{current_strategy_tax} {corporation_strategy_tax} -").safe_evaluate().value

# Leave in the corporation: 12% bracket limits [12,400 - > 50,400] (there are 400 still to go in the 12% bracket)
leave_in_corp = RPNEngine(f"{business_income} 400 -").safe_evaluate().value

print(f"""
Answer:
-------

Current Strategy Tax:     ${current_strategy_tax:,.0f}
Corporation Strategy Tax: ${corporation_strategy_tax:,.0f}
Savings:                  ${savings:,.0f}

a) Current Tax Strategy Savings: ${savings:,.0f}
b) She can still get $400 in income before jumping into the 22% tax bracket, so she should leave ${leave_in_corp:,.0f} in the corporation.
""")

```

Hyundai is considering opening a plant in two neighboring states. One state has a corporate tax rate of 10 percent. If operated in this state, the plant is expected to generate 1,000,000 pretax profit. The other state has a corporate tax rate of 2 percent. If operated in this state, the plant is expected to generate 930,000 of pretax profit. 

- a) Which state should Hyundai choose?  
- b) Why do you think the plant in the state with a lower tax rate would produce a lower before-tax income?

```python
data = {
    "States": ["State_1", "State_2"],
    "Tax Rates": [.10, .02],
    "Pretax Profit": [1000000, 930000],
    "Corporate Tax": [],
    "After-Tax Profit": []
}

for k, v in enumerate(data["Tax Rates"]):
    data["Corporate Tax"].append(data["Pretax Profit"][k] * v)
    data["After-Tax Profit"].append(data["Pretax Profit"][k] - data["Corporate Tax"][k])
df = pd.DataFrame(data)

# Format Dataframe Styles
df["Tax Rates"] = df["Tax Rates"].apply(lambda x: x * 100 if isinstance(x, (int, float)) else x)
df["Tax Rates"] = df["Tax Rates"].apply(lambda x: f"{x:.2f}%" if isinstance(x, (int, float)) else x)
df["Pretax Profit"] = df["Pretax Profit"].map("${:,.0f}".format)
df["Corporate Tax"] = df["Corporate Tax"].map("${:,.0f}".format)
df["After-Tax Profit"] = df["After-Tax Profit"].map("${:,.0f}".format)

display(HTML(df.to_html(index=False))) 
print(f""" 
Answers:
      a) All things being equal, Hyunday should choose State_2 
      which has the lower corporate tax rate of {data["Tax Rates"][1]*100:.0f}% and After-Tax 
      Profit of ${data["After-Tax Profit"][1]:,.0f}.

      b) The lower before-tax profit in State 2 likely reflects 
      and implicit tax. because the 2% rate makes this state attractive, 
      more businesses compete to operate there which drives up costs and 
      drives down what they can charge.  That competition eats into pretax 
      profits.
""")
```

Daniel is considering selling two stocks that have not performed well over recent years. A friend recently informed Daniel that one of his stocks has a special designation, which allows him to treat a loss up to 50,000 on this stock as an ordinary loss rather than the typical capital loss. Daniel figures that he has a loss of 60,000 on each stock. If Daniel’s marginal tax rate is 35 percent and he has 120,000 of other capital gains (taxed at 15 percent), what is the tax savings from the special tax treatment?

```python
marginal_tax_rate = .35
capital_tax_rate = .15

capital_gains = 120000
stock_a = 60000 # Section 1244 Stock
stock_b = 60000
section_1244_limit = 50000

output = {
    "Scenario": ["No election", "With election - capital gains tax", "With election - ordinary loss savings"],
    "Effect": [],
}

# Sceneraio 1: No Section 1244 Loss Election
effect = RPNEngine(f"{capital_gains} {stock_a} - {stock_b} - {capital_tax_rate} *").safe_evaluate().value
output["Effect"].append(effect)

# Scenario 2: With Section 1244 Loss Election
effect = RPNEngine(f"{capital_gains} {stock_a} {section_1244_limit} - - {stock_b} - {capital_tax_rate} *").safe_evaluate().value
output["Effect"].append(effect)

effect = RPNEngine(f"{section_1244_limit} {marginal_tax_rate} *").safe_evaluate().value
output["Effect"].append(-effect) # type: ignore

df = pd.DataFrame(output)
total_row = pd.DataFrame([{"Scenario": "Total Savings", "Effect": df["Effect"].sum()}])
df = pd.concat([df, total_row], ignore_index=True)

df["Effect"] = df["Effect"].map("${:,.0f}".format)

print(f"""The special designation Daniel's friend is talking about is the Section 1244 Stock, which offers 
a special tax treatment available for shares of stock issued by small, domestic corporations. 
Under Section 1244, losses from the sale of these shares can be deducted as ordinary losses up to a maximum of $50,000.

The tax benefit available for Daniel if he chooses the special election is $10,000, see below for more details.
       """)
display(HTML(df.to_html(index=False))) 

print("Note: $-10,000 refers to savings not a liability")

```

Irene is saving for a new car she hopes to purchase either four or six years from now. Irene invests 10,000 in a growth stock that does not pay dividends and expects a 6 percent annual before-tax return (the investment is tax deferred). When she cashes in the investment after either four or six years, she expects the applicable marginal tax rate on long-term capital gains to be 25 percent.

- a) What will be the value of this investment four years from now? Six years from now?
- b) When Irene sells the investment, how much cash will she have after taxes to purchase the new car (four and six years from now)?

```python
investment = -10000
rate_of_return = .06
marginal_tax_rate = .25


data = {
    "Years": [4, 6],
    "Value": [],
    "After-tax": []
    }

for year in data["Years"] :
    fv = RPNEngine(f"{rate_of_return} {year} 0 {investment} fv").safe_evaluate().value
    after_tax = RPNEngine(f"{fv} {fv} {investment} + {marginal_tax_rate} * -").safe_evaluate().value
    data["Value"].append(fv)
    data["After-tax"].append(after_tax)

df = pd.DataFrame(data)
df["Value"] = df["Value"].map("${:,.0f}".format)
df["After-tax"] = df["After-tax"].map("${:,.0f}".format)
display(HTML(df.to_html(index=False))) 

print(f"""
a) The value of the investments:
      Four years from now: ${data["Value"][0]:,.0f}
      Six Years from now:  ${data["Value"][1]:,.0f}

b) The after-tax value of the investments:
      Four years from now: ${data["After-tax"][0]:,.0f}
      Six Years from now:  ${data["After-tax"][1]:,.0f}
      
""")
```

Komiko Tanaka invests 12,000 in LymaBean, Inc. LymaBean does not pay any dividends. Komiko projects that her investment will generate a 10 percent before-tax rate of return. She plans to invest for the long term.

- a) How much cash will Komiko retain, after taxes, if she holds the investment for five years and then sells it when the long-term capital gains rate is 15 percent?
- b) What is Komiko’s after-tax rate of return on her investment in part (a)?
- c) How much cash will Komiko retain, after taxes, if she holds the investment for five years and then sells when the long-term capital gains rate is 25 percent?
- d) What is Komiko’s after-tax rate of return on her investment in part (c)?
- e) How much cash will Komiko retain, after taxes, if she holds the investment for 15 years and then sells when the long-term capital gains rate is 15 percent?
- f) What is Komiko’s after-tax rate of return on her investment in part (e)?


```python
# Lets a function so we don't have to re-write the solution every time.

def after_taxes(investment, rate_of_return, tax_rate, years):
    fv_of_cash = RPNEngine(f"{rate_of_return} {years} 0 {investment} fv").safe_evaluate().value
    tax = RPNEngine(f"{fv_of_cash} {investment} + {tax_rate} *").safe_evaluate().value
    after_tax = RPNEngine(f"{fv_of_cash} {tax} -").safe_evaluate().value

    # After-Tax Rate of Return
    # CAGR = Compounded Annual Growth Rate
    # (After_tax / investment) ^ (1 / N) - 1
    after_tax_return = RPNEngine(f"{after_tax} {-investment} / {1/years} pow 1 -").safe_evaluate().value
    return [after_tax, after_tax_return]
```

```python
# Questions a & b
after_tax_15_pct= after_taxes(-12000, .10, .15, 5)
after_tax_25_pct = after_taxes(-12000, .10, .25, 5)
after_tax_15_years = after_taxes(-12000, .10, .15, 15)
print(f"""
- a) If Komiko's capital gains tax rate is 15%, she will have ${after_tax_15_pct[0]:,.0f} after taxes when she sells in {years} years.
- b) Komiko's after-tax return on her investment is {after_tax_15_pct[1]*100:.2f}%.  
- c) If Komiko's capital gains tax rate is 25%, she will have ${after_tax_25_pct[0]:,.0f} after taxes when she sells in {years} years.
- d) Komiko's after-tax return on her investment is {after_tax_25_pct[1]*100:.2f}%.  
- e) If Komiko's capital gains tax rate is 15%, she will have ${after_tax_15_years[0]:,.0f} after taxes when she sells in {years} years.
- f) Komiko's after-tax return on her investment is {after_tax_15_years[1]*100:.2f}%.  
""") 
```

| | 5 years | Annual After-Tax Rate of Return | 18 Years | Annual After-Tax Rate of Return |
|---|---|---|---|---:|
|Corporate bonds (ordinary interest taxed annually) | 5.75% | | 4.75% | |
|Dividend-paying stock (no appreciation and dividends taxed at 15%) | 3.50% | | 3.50% | |
|Growth stock  | FV = 65,000 | | FV = 140,000 | |
|Municipal bond (tax-exempt) | 3.50% | | 3.50% | |

Complete the two Annual After-Tax Rate of Return columns for each investment and provide investment recommendations for Alan and Alice.

```python
marginal_tax_rate = .32
capital_gains_rate = .15

# 5 Years - Vacation Home
corporate_bonds = RPNEngine(f".0575 1 {marginal_tax_rate} - *").safe_evaluate().value
dividend_stock = RPNEngine(f".0350 1 {capital_gains_rate} - *").safe_evaluate().value
growth_stock_tax = RPNEngine(f"65000 50000 - .15 *").safe_evaluate().value
growth_stock_after_tax = RPNEngine(f"65000 {growth_stock_tax} -").safe_evaluate().value
growth_stock_after_tax_rate = RPNEngine(f"{growth_stock_after_tax} 50000 / {1 / 5} pow 1 -").safe_evaluate().value
municipal_bond = .0350
vacation_home = [corporate_bonds, dividend_stock, growth_stock_after_tax_rate, municipal_bond]

# 18 Years - Cooper's Education
corporate_bonds = RPNEngine(f".0475 1 {marginal_tax_rate} - *").safe_evaluate().value
dividend_stock = RPNEngine(f".0350 1 {capital_gains_rate} - *").safe_evaluate().value
growth_stock_tax = RPNEngine(f"140000 50000 - .15 *").safe_evaluate().value
growth_stock_after_tax = RPNEngine(f"140000 {growth_stock_tax} -").safe_evaluate().value
growth_stock_after_tax_rate = RPNEngine(f"{growth_stock_after_tax} 50000 / {1 / 18} pow 1 -").safe_evaluate().value
municipal_bond = .0350
cooper_education = [corporate_bonds, dividend_stock, growth_stock_after_tax_rate, municipal_bond]

data = {
    "Investment": ["Corporate bonds", "Dividend-paying stock", "Growth stock", "Municipal bond"],
    "5 Years Annual Rate": vacation_home,
    "18 Years Annual Rate": cooper_education,
}
# for k, v in data.items():
#    print(k, v)

df = pd.DataFrame(data)
df["5 Years Annual Rate"] = df["5 Years Annual Rate"].apply(lambda x: x * 100 if isinstance(x, (int, float)) else x)
df["5 Years Annual Rate"] = df["5 Years Annual Rate"].apply(lambda x: f"{x:.2f}%" if isinstance(x, (int, float)) else x)
df["18 Years Annual Rate"] = df["18 Years Annual Rate"].apply(lambda x: x * 100 if isinstance(x, (int, float)) else x)
df["18 Years Annual Rate"] = df["18 Years Annual Rate"].apply(lambda x: f"{x:.2f}%" if isinstance(x, (int, float)) else x)
display(HTML(df.to_html(index=False))) 

print(f"""
Recommendations:
- Vacation Home (5 Years): Growth stock has the highest after-tax return of 4.65%
- Cooper's Education (18 Years): Growth stock has the higest after-tax return of 5.29%
""")
```
