Isabel, a calendar-year taxpayer, uses the cash method of accounting for her sole proprietorship. In late December she received a 20,000 bill from her accountant for consulting services related to her small business. Isabel can pay the 20,000 bill anytime before January 30 of next year without penalty. Assume her marginal tax rate is 37 percent this year and next year, and that she can earn an after-tax rate of return of 12 percent on her investments. When should she pay the 20,000 bill—this year or next?

```python
# This is a Deduction Timing Strategy problem when tax rates are constant, we already know that in this
# kind of situations, the tax payer is better off accelerating deductions into earlier years to maximize
# the present value of tax savings from deductions, but for illustration purposes we will provide support 
# to our conclusion.

category = "expense"
before_tax_amount = 20000
rate_of_return = .12

scenarios = {
    "Scenario_1": [0, .37], # This year
    "Scenario_2": [1, .37]  # Next year
}

timming_strategy(category, before_tax_amount, 0, rate_of_return, scenarios)

print(f"""Since Isabel prefers to pay $12,600 this year, rather 
than $13,393 next year, she chooses this year.""")
```

```python
# This is a timming strategy problem when the tax rate changes.
# As a general rule, deductions should be taken in years when 
# tax rates are higher, and income when tax rates are lower. 
# But that will depend, each scenario should be tested first.

category = "expense"
amount = 20000
rate_of_return = .12

scenarios = {
    "Scenario_1": [0, .32], 
    "Scenario_2": [1, .37]
}

timming_strategy(category, amount, 0, rate_of_return, scenarios)

print(f"""Becase the after-tax cost of the bill is greater this year 
in comparison with next year: 

(Scenario_1: $13,600) - this year
(Scenario_2: $13,393) - next year 
      
Reese chooses to pay next year.""")
```

Geraldo recently won a lottery and chose to receive 100,000 today instead of an equivalent amount in 10 years, computed using an 8 percent rate of return. Today, he learned that interest rates are expected to increase in the future. Is this good news for Geraldo given his decision?

```python
rate_of_return = .08
years = 10
pmt = 0
award = 100000

fv = RPNEngine(f"{rate_of_return} {years} {pmt} {-award} fv").safe_evaluate().value
print(f"""Assumming no inflation, this is good news for Geraldo.
      
If we calculate the future value of ${award:,.0f} in {years} years at a rate of return of {rate_of_return*100:.0f}%, we get ${fv:,.0f}.
This was the rate used to calculate the equivalent future amount Geraldo turned down when he chose the lump sum today.""")

rate_of_return = .12
fv = RPNEngine(f"{rate_of_return} {years} {pmt} {-award} fv").safe_evaluate().value
print(f"""Now that interest rates are expected to rise, this benefits Geraldo in two ways. 
      
      * First, he can invest his ${award:,.0f} at the higher rate and earn a better return than originally expected — for example, 
      at a {rate_of_return*100:.0f}% rate of return, the future value of ${award:,.0f} in {years} years would be ${fv:,.0f}. 
      
      * Second, the future payout he turned down becomes relatively less attractive, since it was calculated using the lower 8% rate; 
      in today's terms, that fixed future amount is now worth less than it would be under the new, higher-rate environment.""")
```

Tawana owns and operates a sole proprietorship and has a 37 percent marginal tax rate. She provides her son, Jonathon, $8,000 a year for college expenses. Jonathon works as a pizza
delivery person every fall and has a marginal tax rate of 15 percent.

- a) What could Tawana do to reduce her family tax burden?
- b) How much pretax income does it currently take Tawana to generate the $8,000 (after taxes) given to Jonathon?
- c) If Jonathon worked for his mother’s sole proprietorship, what salary would she have to pay him to generate $8,000 after taxes (ignoring any Social Security, Medicare, or self- employment tax issues)?
- d) How much money would the strategy in part (c) save?

```python
tawana_mtr = .37 # Tawana's marginal tax rate
jonathon_mtr = .15   # Jonathon's margina tax rate
amount = 8000    #  Needed for college

pretax_income_tawana = RPNEngine(f"{amount} 1 {tawana_mtr} - /").safe_evaluate().value 
pretax_income_jonathon = RPNEngine(f"{amount} 1 {jonathon_mtr} - /").safe_evaluate().value 
savings = RPNEngine(f"{pretax_income_tawana} {pretax_income_jonathon} -").safe_evaluate().value
print(f"""
Answers:

a) Tawana could use an Income Shifting strategy to shift some income to her son who is in a lower tax bracket.

b) It takes ${pretax_income_tawana:,.0f} to generate ${amount:,.0f} after taxes given to Jonathon.

c) She would have to pay him ${pretax_income_jonathon:,.0f} to generate ${amount:,.0f} after taxes.

d) The strategy in part c would save ${savings:,.0f}

""")
```

Moana is a single taxpayer who operates a sole proprietorship. She expects her taxable income next year to be 250,000, of which 200,000 is attributed to her sole proprietorship. Moana is contemplating incorporating her sole proprietorship. Using the 2026 single individual tax brackets, the 2026 corporate tax rate and ignoring any Social Security, Medicare, or self-employment tax issues find out:

- a) How much current tax this strategy could save Moana. 
- b) How much income should be left in the corporation?

```python
status = "SNG"
corporate_tax_rate = .21
total_income = 250000
business_income = 200000
personal_income = RPNEngine(f"{total_income} {business_income} -").safe_evaluate().value

# Current strategy Tax
current_strategy_tax =  OrdinaryTax(status, total_income).compute_tax()

# Corporation strategy tax
personal_tax = OrdinaryTax(status, personal_income).compute_tax()
corporate_tax = RPNEngine(f"{business_income} {corporate_tax_rate} *").safe_evaluate().value
corporation_strategy_tax = RPNEngine(f"{personal_tax} {corporate_tax} +").safe_evaluate().value

# Current Tax Savings
savings = RPNEngine(f"{current_strategy_tax} {corporation_strategy_tax} -").safe_evaluate().value

# Leave in the corporation: 12% bracket limits [12,400 - > 50,400] (there are 400 still to go in the 12% bracket)
leave_in_corp = RPNEngine(f"{business_income} 400 -").safe_evaluate().value

print(f"""
Answer:
-------

Current Strategy Tax:     ${current_strategy_tax:,.0f}
Corporation Strategy Tax: ${corporation_strategy_tax:,.0f}
Savings:                  ${savings:,.0f}

a) Current Tax Strategy Savings: ${savings:,.0f}
b) She can still get $400 in income before jumping into the 22% tax bracket, so she should leave ${leave_in_corp:,.0f} in the corporation.
""")

```

Orie and Jane, husband and wife, operate a sole proprietorship. They expect their taxable
income next year to be 450,000, of which 250,000 is attributed to the sole proprietorship. Orie and Jane are contemplating incorporating their sole proprietorship. Using the 2026 married-joint tax brackets and the corporate tax rate, find out: 

- a) How much current tax this strategy could save Orie and Jane. 
- b) How much income should be left in the corporation?

```python
status = "MFJ" 
corporate_tax_rate = .21
total_income = 450000
business_income = 250000
personal_income = RPNEngine(f"{total_income} {business_income} -").safe_evaluate().value

# Current strategy Tax
current_strategy_tax =  OrdinaryTax(status, total_income).compute_tax()

# Corporation strategy tax
personal_tax = OrdinaryTax(status, personal_income).compute_tax()
corporate_tax = RPNEngine(f"{business_income} {corporate_tax_rate} *").safe_evaluate().value
corporation_strategy_tax = RPNEngine(f"{personal_tax} {corporate_tax} +").safe_evaluate().value

# Current Tax Savings
tax_savings = RPNEngine(f"{current_strategy_tax} {corporation_strategy_tax} -").safe_evaluate().value

# Leave in the corporation: 22% bracket limits [100,800 - > 211,400] (by taking money out they'll be taxed at a higher rate)
leave_in_corp = RPNEngine(f"{business_income} 400 -").safe_evaluate().value

print(f"""
Answer:
-------

Current Strategy Tax:     ${current_strategy_tax:,.0f}
Corporation Strategy Tax: ${corporation_strategy_tax:,.0f}
Savings:                  ${tax_savings:,.0f}

a) Current Tax Strategy Savings: ${tax_savings:,.0f}
b) Since they are currently in the 22% tax bracket they should leave the entire business income in the corporation.

Of course, this is a textbook example that assumes they don't need to take out the money.
""")
```
