- **a) Asia owns stock that is listed on the New York Stock Exchange, and this year the stock increased in value by 20,000.**

**Answer:**

- Economic income: Yes - 20,000 increase in wealth.

- Recognized: No - because it wasn't realized

- **b) Ben sold stock for 10,000 and paid a sales commission of 250. Ben purchased the stock several years ago for 4,000.**

##### **Devon owns 1,000 shares of stock worth 10,000. This year he received 200 additional shares of this stock from a stock dividend. His 1,200 shares are now worth 12,500. Must Devon include the dividend paid in stock in income?**

**Answer:**

**IRC 305**(a) provides the basic rule:

"Gross income does not include the amount of any distribution of the stock of a corporation made by such corporation to its shareholders with respect to its stock".

The stock dividend does not represent a taxable event, hence Devon does not include it in gross income.

- **Answer:** The following will apply to the next set of questions:
    - **IRC 61:** *"Except as otherwise provided in this subtitle, gross income means all income from whatever source derived..."*

    - **Answer:** Yes. Jay must include the 20,000 stolen from his employer.

    - **Answer:** Yes. Keisha must include the 1,000 in the year it was found. Reg 1.61-14 also apply here.

##### **Last year Acme paid Ralph 15,000 to install a new air-conditioning unit at its headquarters building. The air conditioner did not function properly, and this year Acme requested that Ralph return the payment. Because Ralph could not repair one critical part in the unit, he refunded the cost of the repair, 5,000, to Acme.**

- **a) Is Ralph required to include the 15,000 payment he received last year in his gross income from last year?**
    - **Answer:** Yes, Ralph is required to include the 15,000 payment in gross income.

- **b) What are the tax implications of the repayment if Ralph was in the 35 percent tax bracket when he received the 15,000 payment from Acme but he was in the 24 percent tax bracket when he refunded 5,000 to Acme?**

```python
last_year_tax_rate = .35
this_year_tax_rate = .24
this_year_refund = 5000

# Option A: Deduction in the repayment year
option_a = this_year_refund * this_year_tax_rate

# Option B: Take a credit under IRC 1341
option_b = this_year_refund * last_year_tax_rate

relief = max(option_a, option_b)

print(f"""
Under IRC 1341, when a taxpayer included an amount
in gross income in an earlier year because it appeared
to be his right to the money, but later discovers that 
he did not have an unrestricted right to that amount and
repays it, he may receive relief.

Here Ralph has two options:

      Option A: Deduction in the repayment year:   $ {option_a:,.0f}
      Option B: Take a credit under IRC 1341:      $ {option_b:,.0f}

Since Option B offers the best outcome, he chooses Option B.""")
```

##### **Louis files as a single taxpayer. In April of this year he received a 900 refund of state income taxes that he paid last year. How much of the refund, if any, must Louis include in gross income under the following independent scenarios? Assume the standard deduction last year was 13,850.**

```python
refund = 900
standard_deduction = 13850
```

- **a) Last year Louis claimed itemized deductions of 14,100. Louis’s itemized deductions included state income taxes paid of 1,750 and no other state or local taxes.**

```python
itemized_deductions = 14100
excess = itemized_deductions - standard_deduction

print(f"""
Tax Benefit Rule: A state tax refund is included
in gross income this year only to the extent it
produced a tax benefit last year.

(1) Itemized deduction claimed last year:   {itemized_deductions:,.0f}
(2) Standard deduction last year:           {standard_deduction:,.0f}

Excess itemized over standard               {excess:,.0f}

Louis should include $ {excess:,.0f} in gross income
this year.""")
```

- **b) Last year Louis had itemized deductions of 10,800 and he chose to claim the standard deduction. Louis’s itemized deductions included state income taxes paid of 1,750 and no other state or local taxes.**
    - **Answer:** Since Louis chose the standard deduction last year, the state income tax refund received this year is not included in gross income.

- **c) Last year Louis claimed itemized deductions of 15,290. Louis’s itemized deductions included state income taxes paid of 2,750 and no other state or local taxes.**

```python
itemized_deductions = 15290
excess = itemized_deductions - standard_deduction

print(f"""
Tax Benefit Rule: A state tax refund is included
in gross income this year only to the extent it
produced a tax benefit last year.

(1) Itemized deduction claimed last year:   {itemized_deductions:,.0f}
(2) Standard deduction last year:           {standard_deduction:,.0f}

Excess itemized over standard               {excess:,.0f}
State Income Tax Refund                     {refund:,.0f}

Louis should include $ {min(excess, refund):,.0f} in gross income
this year.""")
```

##### **Ralph owns a building that he is trying to lease. Ralph is a calendar-year, cash-method taxpayer and is trying to evaluate the tax consequences of three different lease arrangements. Under lease 1, the building rents for 500 per month, payable on the first of the next month, and the tenant must make a 500 security deposit that is refunded at the end of the lease. Under lease 2, the building rents for 5,500 per year, payable at the time the lease is signed, but no security deposit is required. Under lease 3, the building rents for 500 per month, payable at the beginning of each month, and the tenant must pay a security deposit of 1,000 that is to be applied toward the rent for the last two months of the lease.**

- **a) What amounts are included in Ralph’s gross income this year if a tenant signs lease 1 on December 1 and makes timely payments under that lease?**
    - **Answer:** If Ralph chooses Lease 1, he recognizes 0 dollars this year.  The security deposit is refundable and not included in income, the first payment of the lease is due on January 1st of next year.  In conclusion, he doesn't recognize any income relating to this lease.

- **b) What amounts are included in Ralph’s gross income this year if the tenant signs lease 2 on December 31 and makes timely payments under that lease?**
    - **Answer:** Lease 2 explicitly states that the lease payment of 5,500 is payable at the time the lease is signed.  Since it was signed on December 31, Ralph should recognize the whole 5,500 this year.

- **c) What amounts are included in Ralph’s gross income this year if the tenant signs lease 3 on November 30 and makes timely payments under that lease?**

```python
rent = 500
months = 1 # December
deposit = 1000
gross_income = (rent * months) + deposit

print(f"""
Ralph should recognize ${gross_income:,.0f} as gross income this year: 
${rent:,.0f} for December rent, plus the ${deposit:,.0f} deposit. Because
the deposit is earmarked to prepay rent for the last two 
months of the lease, it is treated as advance rent rather 
than a true security deposit, and advance rent is included 
in gross income when received.""")
```

##### **Anne purchased an annuity from an insurance company that promised to pay her 20,000 per year for the next 10 years. Anne paid 145,000 for the annuity, and in exchange she will receive 200,000 over the term of the annuity.**

```python
periods = 10     # Years
annuity = 145000 # Cost
payments = 20000 # Payments per year
expected_return = payments * periods
exclusion_ratio = annuity / expected_return
return_of_capital = annuity / periods # Per month
```

- **a) How much of the first 20,000 payment should Anne include in gross income?**

```python
gross_income = payments - return_of_capital
print(f"""
Summary:
--------
Investment in Annuity    $ {annuity:,.0f} 
Annual Payments            {20000:,.0f}
Expected Return            {expected_return:,.0f}
Exclusion Ratio            {exclusion_ratio}
Years                      {periods}

Anne should include ${gross_income:,.0f} in gross income
for the first payment.

Calculation:
Included in gross income = payments * (1 - exclusion ratio)
""")
```

- **b) How much income will Anne recognize over the term of the annuity?**

```python
gross_income = (payments * periods) - annuity
print(f"""
Anne will recognize ${gross_income:,.0f} over the
term of the annuity.

Calculation:
gross income = (annual payments * years) - investment in annuity""")
```
