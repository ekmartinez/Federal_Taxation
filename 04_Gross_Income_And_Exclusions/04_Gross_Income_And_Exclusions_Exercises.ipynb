##### **For the following independent cases, determine whether economic income is present and, if so, whether it must be included in gross income (i.e., is it realized and recognized for tax purposes?)**

- **a) Asia owns stock that is listed on the New York Stock Exchange, and this year the stock increased in value by 20,000.**

**Answer:**

- Economic income: Yes - 20,000 increase in wealth.
- Realized: No - no sale, exchange, or other triggering event.
- Recognized: No - because it wasn't realized

- **b) Ben sold stock for 10,000 and paid a sales commission of 250. Ben purchased the stock several years ago for 4,000.**

```python
realized = 10000
cost = 4000
commission = 250
recognized = realized - cost - commission

print(f"""There is Economic Benefit:
      
    Amount realized ({realized:,.0f} realized - {commission:,.0f} commission): {realized - commission:,.0f}
    Basis:                                             ({cost:,.0f})
                                                       -------
    Realized & Recognized LT Capital Gain:              {recognized:,.0f}""")
```

##### **Devon owns 1,000 shares of stock worth 10,000. This year he received 200 additional shares of this stock from a stock dividend. His 1,200 shares are now worth 12,500. Must Devon include the dividend paid in stock in income?**

**Answer:**

**IRC 305**(a) provides the basic rule:

"Gross income does not include the amount of any distribution of the stock of a corporation made by such corporation to its shareholders with respect to its stock".

The stock dividend does not represent a taxable event, hence Devon does not include it in gross income.

##### **For the following independent cases, determine whether economic income is present and, if so, whether it must be included in gross income. Identify a tax authority that supports your analysis.**

- **Answer:** The following will apply to the next set of questions:
    - **IRC 61:** *"Except as otherwise provided in this subtitle, gross income means all income from whatever source derived..."*

- **a) Hermione discovered a gold nugget (valued at 10,000) on her land.**

    - **Answer:** Yes. Hermione must include the value of the gold nugget found (10,000). This falls under the "treasure trove" doctrine, specifically Reg.1.61-14, which explicitly states treasure trove is includible in gross income for the year it's reduced to undisputed possession, at its fair market value.

- **b) Jay embezzled 20,000 from his employer and has not yet been apprehended.**

    - **Answer:** Yes. Jay must include the 20,000 stolen from his employer.

- **c) Keisha found 1,000 inside an old dresser. She purchased the dresser at a discount furniture store at the end of last year and found the money after the beginning of the new year. No one has claimed the money.**

    - **Answer:** Yes. Keisha must include the 1,000 in the year it was found. Reg 1.61-14 also apply here.

##### **Although Hank is retired, he is an excellent handyman and often works part time on small projects for neighbors and friends. Last week his neighbor, Mike, offered to pay Hank 500 for minor repairs to his house. Hank completed the repairs in December of this year. Hank uses the cash method of accounting and is a calendar-year taxpayer. Compute Hank’s gross income for this year from each of the following alternative transactions:**

- **a) Mike paid Hank 200 in cash in December of this year and promised to pay the remaining 300 with interest in three months.**
    - **Answer:** Hank's gross income would be the 200 received in December.  Then, on the next tax year he would report the remaining 300 and whatever interest generated as interest income.

- **b) Mike gave Hank tickets in December to the big game in January. The tickets have a face value of 50, but Hank could sell them for 400. Hank went to the game with his son.**
    - **Answer:** Hank's gross income would include 400, the market value of the ticket.

- **c) Mike bought Hank a new set of snow tires and gave them to Hank in December. The tires typically sell for 500, but Mike bought them on sale for 450.**
    - **Answer:** Hank should report the market value of the snow tires as gross income: 500.

##### **Last year Acme paid Ralph 15,000 to install a new air-conditioning unit at its headquarters building. The air conditioner did not function properly, and this year Acme requested that Ralph return the payment. Because Ralph could not repair one critical part in the unit, he refunded the cost of the repair, 5,000, to Acme.**

- **a) Is Ralph required to include the 15,000 payment he received last year in his gross income from last year?**
    - **Answer:** Yes, Ralph is required to include the 15,000 payment in gross income.

- **b) What are the tax implications of the repayment if Ralph was in the 35 percent tax bracket when he received the 15,000 payment from Acme but he was in the 24 percent tax bracket when he refunded 5,000 to Acme?**

```python
last_year_tax_rate = .35
this_year_tax_rate = .24
this_year_refund = 5000

# Option A: Deduction in the repayment year
option_a = this_year_refund * this_year_tax_rate

# Option B: Take a credit under IRC 1341
option_b = this_year_refund * last_year_tax_rate

relief = max(option_a, option_b)

print(f"""
Under IRC 1341, when a taxpayer included an amount
in gross income in an earlier year because it appeared
to be his right to the money, but later discovers that 
he did not have an unrestricted right to that amount and
repays it, he may receive relief.

Here Ralph has two options:

      Option A: Deduction in the repayment year:   $ {option_a:,.0f}
      Option B: Take a credit under IRC 1341:      $ {option_b:,.0f}

Since Option B offers the best outcome, he chooses Option B.""")
```
