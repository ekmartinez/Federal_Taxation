##### **Devon owns 1,000 shares of stock worth 10,000. This year he received 200 additional shares of this stock from a stock dividend. His 1,200 shares are now worth 12,500. Must Devon include the dividend paid in stock in income?**

**Answer:**

**IRC 305**(a) provides the basic rule:

"Gross income does not include the amount of any distribution of the stock of a corporation made by such corporation to its shareholders with respect to its stock".

The stock dividend does not represent a taxable event, hence Devon does not include it in gross income.

##### **Anne purchased an annuity from an insurance company that promised to pay her 20,000 per year for the next 10 years. Anne paid 145,000 for the annuity, and in exchange she will receive 200,000 over the term of the annuity.**

```python
periods = 10     # Years
annuity = 145000 # Cost
payments = 20000 # Payments per year
expected_return = payments * periods
exclusion_ratio = annuity / expected_return
return_of_capital = annuity / periods # Per month
```

- **a) How much of the first 20,000 payment should Anne include in gross income?**

```python
gross_income = payments - return_of_capital
print(f"""
Summary:
--------
Investment in Annuity    $ {annuity:,.0f} 
Annual Payments            {20000:,.0f}
Expected Return            {expected_return:,.0f}
Exclusion Ratio            {exclusion_ratio}
Years                      {periods}

Anne should include ${gross_income:,.0f} in gross income
for the first payment.

Calculation:
Included in gross income = payments * (1 - exclusion ratio)
""")
```

- **b) How much income will Anne recognize over the term of the annuity?**

```python
gross_income = (payments * periods) - annuity
print(f"""
Anne will recognize ${gross_income:,.0f} over the
term of the annuity.

Calculation:
gross income = (annual payments * years) - investment in annuity""")
```

##### **Priya works for the Shine Company, a retailer of upscale jewelry. How much taxable income does Priya recognize under the following scenarios?**

**Note:**

Before answering this question, let's look at the tax consequences of bargain purchases between employers and its employees. The general rule is that bargain purchases by an employee from an employer create taxable compensation income to the employee, the tax law does provide a limited exclusion for employee bargain purchases. Specifically, employees may exclude:

- (a) A discount on employer-provided goods as long as the discount does not exceed the employer’s gross profit percentage on all property offered for sale to customers.

- (b) Up to 20 percent employer-provided discounts on services. Discounts in excess of these amounts are taxable as compensation.

- **a) Priya buys a diamond ring from Shine Company for 10,000 (normal sales price, 14,000; Shine Company’s gross profit percentage is 40 percent).**
