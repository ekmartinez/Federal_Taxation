**Example 5-1**

In April, Gram used part of the life insurance proceeds she received from Gramps’s death to purchase 50 shares in Acme Corporation for 30 per share. From April to the end of December, the value of the shares fluctuated between 40 and 25, but on December 31, the shares were worth 35. If Gram does not sell the shares, how much income from her stockholdings in Acme Corporation does she realize for the year?

```python
beginning_value = 50 * 30
ending_value = 50 * 35
realized = ending_value - beginning_value

print(f"""
Investment in Acme at April 30:      $ {beginning_value:,.0f}
Investment in Acme at December 31:   $ {ending_value:,.0f}
                                     -------
Gain Realized:                       $   {realized:,.0f}
Gain Recognized:                     $     {0:,.0f}
""")
```

**Answer:** 0. Unless Gram sells the stock, she does not enter into a transaction resulting in a measurable change of property rights with a second party. Thus, she does not realize income even though she experienced an economic benefit from the appreciation of the stock from 30 per share to 35 per share.

**Example 5-3**

In 2023 Courtney paid 3,500 in Ohio state income taxes, and she included this payment with her other itemized deductions when she filed her federal income tax return in March of 2024. Courtney filed her 2023 federal return as a head of household and claimed 22,900 of itemized deductions (including 3,500 state income taxes, 2,500 of real estate taxes, 8,000 of mortgage interest expense, and 8,900 of charitable contributions). She also filed an Ohio state income tax return in March of 2024 and discovered she only owed 3,080 in Ohio income tax for 2023. Hence, Courtney received a 420 refund of her Ohio income tax in June of 2024.

Applying the tax benefit rule, how much of the 420 refund, if any, is Courtney required to include in her gross income for 2024 given that the standard deduction for head of household filing status in 2023 was 20,800?

**Answer:** All 420. Courtney is required to include the entire refund in her 2024 gross income because her itemized deduction for the 420 of state income taxes that she overpaid last year reduced her taxable income by 420. Accordingly, because she received a tax benefit (deduction) for the entire 420 overpayment, she must include it all in gross income for 2024. See the following for the calculation of the amount of tax benefit Courtney received from the 420 overpayment of taxes in 2023.

| Deduction | Amount | Explanation |
| --- | :---: | --- |
| (1) Itemized deductions | 22,900 | |
| (2) 2023 standard deduction | 20,800 | Head of household filing status |
| (3) Reduction in taxable income | 22,900 | Greater of (1) and (2) |
| (4) Itemized deductions adjusted for 420 refund | 22,480 | (1) - 420 |
| (5) Reduction in taxable income after adjustment | 22,480 | Greater of (2) and (4) |
| **Taxable benefit due to prior deduction of 420 refund** | **420** | (3) - (5) |

**What if:** Let’s consider alternative fact patterns provided in Scenarios A and B to further illustrate the application of the tax benefit rule.

How much of the 420 refund would Courtney include in her 2024 gross income in Scenarios A and B?

- **Scenario A:** In 2023 Courtney’s itemized deductions, including 3,500 in state taxes, were 15,500.
    - **Answer Scenario A:** 0. As computed below, Courtney received 0 tax benefit from the $420 tax overpayment in 2023, so she need not include any of the refund in her 2024 gross income.

- **Scenario B:** In 2023 Courtney’s itemized deductions, including 3,500 in state taxes, were 21,150.
    - **Answer Scenario B:** 350. As computed below, Courtney received a $350 tax benefit (reduction in taxable income) from the 420 state tax overpayment in 2023, so she must include 350 of the refund in her 2024 gross income.

| Deduction | Scenario A | Scenario B | Explanation
| --- | :---: | :---: | --- |
| (1) Itemized deductions | 15,500 | 21,500 | |
| (2) 2023 standard deduction | 20,800 | 20,800 | Head of household filing status |
| (3) Reduction in taxable income | 20,800 | 21,500 | Greater of (1) and (2) |
| (4) Adjusted itemized deductions | 15,080 | 20,730 | (1) - 420 |
| (5) Reduction in taxable income adjusted for 420 | 20,800 | 20,800 | Greater of (2) and (4) |
| **Tax benefit due to prior deduction** | **0** | **350** | (3) - (5) |

- **What if:** In 2023 Courtney’s itemized deductions were 21,150 and included 10,000 of state taxes (10,500 total state taxes, limited by tax law to 10,000). We discuss the 10,000 state and local tax limitation in the Individual Deductions chapter. How much of the 420 refund would Courtney include in her 2024 gross income?
     - **Answer:** 0. Courtney received 0 tax benefit from the 420 tax overpayment in 2023. Her itemized deduction for state taxes adjusted for the 420 refund would still be 10,000 (10,500 − 420 = 10,080, limited to 10,000) and her total itemized deductions would still be 21,150. So, she need not include any of the refund in her 2024 gross income.

**Example 5-18**

Gram received $7,200 of Social Security benefits this year. Suppose that Gram’s modified AGI is $16,000. What amount of the Social Security benefits must Gram include in her gross income?

```python
social_security = 7200
modified_agi = 16000
taxable_ss = TaxableSocialSecurity(modified_agi, social_security).single()
print(f"Answer: ${taxable_ss:,.0f}")
```

**Answer:** 0. Gram’s modified AGI plus one-half of her Social Security benefits is 19,600 [16,000 modified AGI + (7,200 Social Security benefits × 50%)], which is below 25,000. Hence, Gram may exclude the entire 7,200 of Social Security income from gross income.

**What if:** Assume that Gram received 7,200 of Social Security benefits this year and that her modified AGI is 26,400. What amount of benefits must Gram include in her gross income?

```python
social_security = 7200
modified_agi = 26400
taxable_ss = TaxableSocialSecurity(modified_agi, social_security).single()
print(f"Answer: ${taxable_ss:,.0f}")
```

**Answer:** 2,500. Gram is single and her modified AGI + 50 percent of Social Security benefits falls between 25,000 and 34,000. Her taxable Social Security benefits are the lesser of (a) 50 percent of the Social Security benefits (7,200 × 50% = 3,600) or (b) 50 percent of [26,400 modified AGI + 3,600 (50 percent of Social Security benefits) − 25,000] = 2,500. Thus, her taxable Social Security benefits are 2,500.

**What if:** Assume that Gram received 7,200 of Social Security benefits this year and that her modified AGI is 31,400. What amount of benefits must Gram include in her gross income?

```python
social_security = 7200
modified_agi = 31400
taxable_ss = TaxableSocialSecurity(modified_agi, social_security).single()
print(f"Answer: ${taxable_ss:,.0f}")
```

**Answer:** 4,450. Gram is single and her modified AGI + 50 percent of Social Security benefits exceeds 34,000. Her taxable Social Security benefits are the lesser of (a) 85 percent of Social Security benefits (7,200 × 85% = 6,120) or (b) 85 percent of [31,400 modified AGI + 3,600 (50 percent of Social Security benefits) − 34,000] = 850, plus the lesser of (i) 4,500 or (ii) 50 percent of Social Security benefits (50% × 7,200 = 3,600). This calculation simplifies to the lesser of (a) 6,120 or (b) 4,450 [850 + lesser of (i) 4,500 or (ii) 3,600]. Thus, Gram’s taxable Social Security benefits are 4,450.

**Example 5-29**

Assume Courtney is considering a transfer to EWD’s overseas affiliate. If she transfers, she anticipates she will earn approximately 140,000 in salary. How much of her expected 140,000 annual salary will Courtney be allowed to exclude from her gross income, assuming she meets the residency requirements?

```python
foreign_income = 140000
maximum_exclusion = 126500       # 2024 Inflation Indexed amount
exclusion = min(foreign_income, maximum_exclusion)
print(f"Answer: ${exclusion:,.0f}")
```


**Answer:** She is eligible to exclude 126,500 of her 140,000 of compensation from U.S. taxation in 2024. However, her entire 140,000 salary may be subject to the foreign country’s income tax.


**What if:** Assume Courtney decides to transfer, but she expects her 12-month qualifying period to include only 200 days in the first year and 140 days in the second year.
```python
expected_salary = 77000
first_year_days = 200
days_in_year = 365
exclusion = maximum_exclusion * (first_year_days / days_in_year)
first_year_exclusion = min(expected_salary, exclusion)
print(f"Answer: ${first_year_exclusion:,.0f}")
```


**Answer:** 69,315 [126,500 full exclusion × 200/365 (days in foreign country/days in year)].
