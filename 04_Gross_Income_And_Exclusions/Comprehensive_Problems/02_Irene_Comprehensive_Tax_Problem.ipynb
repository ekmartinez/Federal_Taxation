##### **Irene is disabled and receives payments from a number of sources. The interest payments are from bonds that Irene purchased over past years and a disability insurance policy that Irene purchased herself. Calculate Irene’s gross income.**

| Description | Amount |
| --- | :---: |
| Interest, bonds issued by City of Austin, Texas | 2,000 |
| Social Security benefits | 8,200 |
| Interest, U.S. Treasury bills | 1,300 |
| Interest, bonds issued by Ford Motor Company | 1,500 |
| Interest, bonds issued by City of Quebec, Canada | 750 |
| Disability insurance benefits | 19,500 |
| Distributions from qualified pension plan | 5,400 |

To solve this problem we need to verify if 85% of social security benefits should be added to gross income.  To do this, we first need to arrive at Adjusted Gross Income, from there we should be able to compute Modified Adjusted Gross Income (MAGI).

```python
city_of_austin_interest = 2000
social_security = 8200
treasury_bills = 1300
ford_motor = 1500
city_of_quebec = 750
disability_insurance = 19500
pension = 5400

agi = treasury_bills + ford_motor + pension
magi = agi + city_of_austin_interest + city_of_quebec
taxable_social_security = TaxableSocialSecurity(magi, social_security).single()
print(f"Taxable Social Security: $ {taxable_social_security:,.2f}")
```

Now we know that Irene should not include 85% of Social Security in her gross income.  We arrive at gross_income as follows:

```python
gross_income = treasury_bills + ford_motor + city_of_quebec + pension

print(f"""
Irene's Gross Income
--------------------
Interest from treasury bills:       {treasury_bills:,.0f}
Interest from Ford Motor:           {ford_motor:,.0f}
Interest from City of Quebec:         {city_of_quebec:,.0f}
Qualified pension:                  {pension:,.0f}
                                    -----
Gross Income:                       {gross_income:,.0f}

Note: Since the disability was aquired and payed by Irene, 
she does not include the proceeds in gross income.""")
```
