##### **Charlie was hired by Ajax this year as a corporate executive and a member of the board of directors. During the current year, Charlie received the following payments or benefits paid on his behalf.**

| Description | Amount |
| --- | ---: |
| Salary payments | 92,000 |
| Contributions to qualified pension plan | 10,200 |
| Qualified health insurance premiums | 8,400 |
| Year-end bonus | 15,000 |
| Annual director's fee | 10,000 |
| Group-term life insurance premiums (face value = 40,000) | 750 |
| Whole life insurance premiums (face value = 100,000) | 1,420 |
| Disability premiums (Non-taxable benefit) | 4,350 |

- **a) Charlie uses the cash method and calendar year for tax purposes. Calculate Charlie’s gross income for the current year.**

```python
struct = {
    "salary" : 92000,
    "year_end_bonus" : 15000, 
    "directors_fees" : 10000,
    "life_insurance_premiums" : 1420
}
gross_income = 0
for v in struct.values():
    gross_income += v

print(f"""
Salary payments:                   {struct["salary"]:,.0f}
Year-end bonus:                    {struct["year_end_bonus"]:,.0f}
Annual director's fee:             {struct["directors_fees"]:,.0f}
Whole life insurance premiums:      {struct["life_insurance_premiums"]:,.0f}
                                  -------
Gross income:                   $ {gross_income:,.0f}

Charlie's gross income for the year is: $ {gross_income:,.0f}""")
```

- **b) Suppose that Ajax agrees to pay Charlie an additional 100,000 once Charlie completes five years of employment. Will this agreement alter Charlie’s gross income this year relative to your part (a) answer? Explain.**
    - **Answer:** This agreement won't alter Charlie's gross income for the year, Charlie is a cash basis taxpayer, thus, he does not recognize income until received.

- **c) Suppose that in exchange for his promise to remain with the firm for the next four years, Ajax paid Charlie four years of director’s fees in advance. Will this arrangement alter Charlie’s gross income this year relative to your part (a) answer? Explain.**
    - **Answer:**  In this case Charlie's gross income will increase by the amount of the advance.  Because Charlie is a cash basis taxpayer, he recognizes income when received, so the four year director's fees in advance will increase his gross income for the year.

- **d) Assume that in lieu of a year-end bonus Ajax transferred 500 shares of Bell stock to Charlie as compensation. Further assume that the stock was listed at 35 per share and Charlie would sell the shares by year-end, at which time he expected the price to be 37 per share. Will this arrangement alter Charlie’s gross income this year relative to your part (a) answer? Explain.**

```python
stock = 500 * 35
gross_income = gross_income + stock - struct["year_end_bonus"]

print(f"""
Charlie should include the value of the investment received 
as compensation in gross income ($ {stock:,.0f}). As result
his gross income will increase to ${gross_income:,.0f} for the year.

Gross income before stock compensation:         {gross_income - stock:,.0f}
Stock compensation:                              {stock:,.0f}
                                                -------
Gross income after stock compensation:          {gross_income:,.0f}""")

sale_of_stock = 500 * 37
capital_gain = sale_of_stock - stock
gross_income += capital_gain

print(f"""
If he chooses to sell his investment at year end at the expected
price of $37.00, then he will have a short-term capital 
gain that would increase his gross income to: $ {gross_income:,.0f}.

Gross income before short-term capital gain:    {gross_income - capital_gain:,.0f}
Capital gain:                                     {capital_gain:,.0f}
                                                -------
Gross income after capital gain:                {gross_income:,.0f}""")
```

- **e) Suppose that in lieu of a year-end bonus Ajax made Charlie’s house payments (a total of 23,000). Will this arrangement alter Charlie’s gross income this year relative to your part (a) answer? Explain.**

```python
house_payment = 23000
gross_income = 118420 - 15000 + house_payment

print(f"""
This arrangement will increase Charlie's gross income
by $ {house_payment:,.0f}, bringing gross income up to $ {gross_income:,.0f}.

Gross income before house payments:            {gross_income - house_payment:,.0f}
House payments:                                 {house_payment:,.0f}
                                               -------
Gross income after house payments:             {gross_income:,.0f}""")
```
