##### **Simpson, age 45, is a single individual who is employed full time by Duff Corporation. This year Simpson reports AGI of 50,000 and has incurred the following medical expenses:**

| Description | Amount |
| --- | :---: |
| Dentist charges | 900 |
| Physician charges | 1,800 |
| Optical charges | 500 |
| Cost of eyeglasses | 300 |
| Hospital charges | 2,100 |
| Prescription drugs | 250 |
| Over-the-counter drugs (not insulin) | 450 |
| Medical insurance premiums (not through an exchange) | 775 |

- **a) Calculate the amount of medical expenses that will be included with Simpson’s itemized deductions after any applicable limitations.**

```python
agi = 50000

medical_expenses = {
    "Dentist charges": 900,
    "Physician charges": 1800,
    "Optical charges": 500,
    "Cost of eyeglasses": 300,
    "Hospital charges": 2100,
    "Prescription drugs": 250,
    "Medical Insurance Premiums": 775
}

medical_expenses = sum(medical_expenses.values())
floor_limitation = agi * .075
itemized_deduction = medical_expenses - floor_limitation
print(f"Simpson will include {itemized_deduction:,.0f} of itemized deductions.\n")

print(f"{"Qualified Medical Expenses":<26} | {medical_expenses:>8,.0f}")
print(f"{"Floor limitation":<26} | {floor_limitation:>8,.0f}")
print(f"{"-"*25:<26} | {"-"*8:>8}")
print(f"{"Itemized Deduction":<26} | {itemized_deduction:>8,.0f}")
```

- **b) Suppose that Simpson was reimbursed for 250 of the physician’s charges and 1,200 for the hospital costs. Calculate the amount of medical expenses that will be included with Simpson’s itemized deductions after any applicable limitations.**

```python
reimbursements = 250 + 1200
net_medical_expenses = medical_expenses - reimbursements
itemized_deduction = net_medical_expenses - floor_limitation

print(f"Simpson will include {itemized_deduction:,.0f} of itemized deductions.\n")

print(f"{"Qualified Medical Expenses":<32} | {medical_expenses:>8,.0f}")
print(f"{"Less: Reimbursements":<32} | {reimbursements:>8,.0f}")
print(f"{"-"*32:<32} | {"-"*8:>8}")
print(f"{"Net Qualified Medical Expenses":<32} | {net_medical_expenses:>8,.0f}")
print(f"{"Floor limitation":<32} | {floor_limitation:>8,.0f}")
print(f"{"-"*32:<32} | {"-"*8:>8}")
print(f"{"Itemized Deduction":<32} | {itemized_deduction:>8,.0f}")
```

##### **Calvin reviewed his canceled checks and receipts this year (2024) for charitable contributions, which included an antique painting and IBM stock. He has owned the IBM stock and the painting since 2005.**

| Donee | Item | Cost | FMV |
| --- | --- | ---: | ---: |
| Hobbs Medical Center | IBM stock | 5,000 | 22,000 |
| State Museum | Antique painting | 5,000 | 3,000 |
| A needy family | Food and clothes | 400 | 250 |
| United Way | Cash | 8,000 | 8,000 |

Calculate Calvin’s charitable contribution deduction and carryover (if any) under the following circumstances.

- **a) Calvin’s AGI is 100,000.**

```python
agi = 100000
remaining_agi = agi

cash = 8000
ordinary_property = 3000
cap_gain_property = 22000

cash_limit = agi * .60
ordinary_limit = ordinary_property * .50
cap_gain_limit = agi * .30

cash_deduction = min(cash, cash_limit, remaining_agi)
cash_carryforward = cash - cash_deduction
remaining_agi -= cash_deduction

ordinary_deduction = min(ordinary_property, ordinary_limit, remaining_agi)
ordinary_carryforward = ordinary_property - ordinary_deduction
remaining_agi -= ordinary_deduction

cap_gain_deduction = min(cap_gain_property, cap_gain_limit, remaining_agi)
cap_gain_carryforward = cap_gain_property - cap_gain_deduction
remaining_agi -= cap_gain_deduction

total_deductions = cash_deduction + ordinary_deduction + cap_gain_deduction
total_carryforward = cash_carryforward + ordinary_carryforward + cap_gain_carryforward

results = {
    "Property": ["Cash contributions", "Ordinary contributions", "Capital gain contributions"],
    "Deduction": [cash_deduction, ordinary_deduction, cap_gain_deduction],
    "Carryforward": [cash_carryforward, ordinary_carryforward, cap_gain_carryforward],
}

df = pd.DataFrame(results)
total = pd.DataFrame(([{"Property": "Total", "Deduction": total_deductions, "Carryforward": total_carryforward}]))
df = pd.concat([df, total], ignore_index=True)

df["Deduction"] = df["Deduction"].map("{:,.0f}".format)
df["Carryforward"] = df["Carryforward"].map("{:,.0f}".format)

display(HTML(df.to_html(index=False)))
```
