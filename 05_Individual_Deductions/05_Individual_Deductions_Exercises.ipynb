##### **Clem is married and is a skilled carpenter. Clem’s wife, Wanda, works part time as a substitute elementary school teacher. Determine the amount of Clem’s expenses that are deductible for AGI this year (if any) under the following independent circumstances:**

- a) Clem is self-employed, and this year he incurred 525 in expenses for tools and supplies related to his job. Because neither was covered by a qualified health plan, Clem paid health insurance premiums of 3,600 to provide coverage for himself and Wanda (not through an exchange).
    - **Answer:** Clem would be able to deduct the 525 business expense as a for AGI (above the line) and 3,600 of insurance premiums as an above the line deduction, for a total of 4,125 For AGI deduction for the year.


- b) Clem and Wanda own a garage downtown that they rent to a local business for storage. This year they incurred expenses of 1,250 in utilities and 780 in depreciation.
    - **Answer:** Clem and Wanda would be able to deduct 1,250 utilities expense and 780 depreciation expense for a total of 2,030 for AGI deduction.


- c) Clem paid self-employment tax of 15,300 (the employer portion is 7,650), and Wanda had 3,000 of Social Security taxes withheld from her pay.
    - **Answer:** Clem would be able to deduct the employer portion of the self-employment tax (7,650) as a For AGI deduction.  Wanda's 3,000 are not deductible.


- d) Clem paid 45 to rent a safety deposit box to store his coin collection. Clem has collected coins intermittently since he was a child, and he expects to sell his collection when he retires.
    - **Answer:** Clem may not deduct the payment of 45 to rent a safety deposit related to his hobby expenses.

##### **Don Juan, a single taxpayer, is the sole owner of DJ’s Inc., an S corporation. In 2024, DJ’s Inc. incurred a massive 600,000 business loss, all of which is allocable to Don Juan as the sole shareholder. Assume that the 600,000 loss is not limited by the basis, at-risk, or passive loss rules and that Don Juan has no other business income or business losses. How much of the 600,000 loss will Don Juan be able to deduct this year? What happens to any loss not deducted this year?**

```python
business_loss = 600000
business_loss_limit = 305000 # Single taxpayers 2024 threshold
deductible = min(business_loss, business_loss_limit)
excess = business_loss - deductible

print(f"""Assuming that the $600,000 loss is not limited by basis, at-risk or 
passive loss rules and that Don Juan has no other business income or business
losses:\n""")

print(f"{"Business loss":<26} | {business_loss:>8,.0f}")
print(f"{"Business loss limitation":<26} | {business_loss_limit:>8,.0f}")
print(f"{"-"*26:<26} | {"-"*8:>8}")
print(f"{"Deductible business loss":<26} | {deductible:>8,.0f}")

print(f"\nDon Juan may carry forward to subsequent years the {excess:,.0f} excess business loss.")
```

##### **Smithers is a self-employed individual who earns 30,000 per year in self-employment income. Smithers pays 2,200 in annual health insurance premiums (not through an exchange) for his own medical care. In each of the following situations, determine the amount of the deductible health insurance premium for Smithers before any AGI floor limitation.**

- **a) Smithers is single and the self-employment income is his only source of income.**
    - **Answer:** Smithers may claim the 2,200 in annual health insurance premiums as a For AGI deduction.

- **b) Smithers is single, but besides being self-employed, Smithers is also employed part time by SF Power Corporation. This year Smithers elected not to participate in SF’s health plan.**
    - **Answer:** In this case, Smithers would not be able to claim health insurance premiums as a For AGI deduction because he was eligible to participate in his employer's health insurance plan but decided not to.

- **c) Smithers is self-employed and is also married. Smithers’s spouse, Samantha, is employed full time by SF Power Corporation and is covered by SF’s health plan. Smithers is not eligible to participate in SF’s health plan.**
    - **Answer:** Smithers may claim the 2,200 in annual health insurance premiums as a For AGI deduction, because he is not eligible to participate in his wife's employer provided health insurance plan.

- **d) Smithers is self-employed and is also married. Smithers’s spouse, Samantha, is employed full time by SF Power Corporation and is covered by SF’s health plan. Smithers elected not to participate in SF’s health plan.**
    - **Answer:** In this case, Smithers would not be able to claim health insurance premiums as a For AGI deduction because he was eligible to participate in his wife's employer provided health insurance plan but decided not to.

##### **Hardaway earned 100,000 of compensation this year. He also paid (or had paid for him) 3,000 of health insurance (not through an exchange). What is Hardaway’s AGI in each of the following situations? (Ignore the effects of Social Security and self-employment taxes.)**

- **a) Hardaway is an employee, and his employer paid Hardaway’s 3,000 of health insurance for him as a nontaxable fringe benefit. Consequently, Hardaway received 97,00 of taxable compensation and 3,000 of nontaxable compensation.**

```python
agi = 97000
print(f"Adjusted Gross Income: $ {agi:,.0f}")
print(f"\nThe employer provided health insurance premiums are excluded from gross income.")
```

- **b) Hardaway is a self-employed taxpayer, and he paid 3,000 of health insurance himself. He is not eligible to participate in an employer-sponsored plan.**

```python
income = 100000
health_insurance = 3000
agi = income - health_insurance

print(f"{"Income Earned":<30} | {income:>8,.0f}")
print(f"{"Health Insurance Premiums":<30} | {health_insurance:>8,.0f}")
print(f"{"-"*30:<30} | {"-"*8:>8}")
print(f"{"Adjusted Gross Income":<30} | {agi:>8,.0f}")
print(f"""\nBecause Hardaway is self-employed and he is not eligible to participate
in his employer's health insurance plan, he would be able to deduct {health_insurance:,.0f}
as a For agi deduction.""")
```

##### **Rocco operates a beauty salon as a sole proprietorship. Rocco also owns and rents an apartment building. In 2024, Rocco had the following income and expenses. | Depreciation on apartment building | 12,900 |
