**Example 6-2**

Besides being employed by EWD, Courtney is also a self-employed architectural consultant (a business activity). This year her consulting activity generated 19,500 of revenue and incurred 1,500 in expense (primarily travel and transportation expenses). How does she report the revenue and deductions from the activity?

````python
business_revenue = 19500
business_expenses = 1500
net_income = business_revenue - business_expenses
print(f"{"Business Revenue":18} | {business_revenue:>6,.0f}")
print(f"{"Business Expenses":<18} | {business_expenses:>6,.0f}")
print(f"{"-"*18:<18} | {"-"*6:<6}")
print(f"{"Net Income":<18} | {net_income:>6,.0f}")
````

**Answer**: Courtney reports the 19,500 of revenue and deducts the $1,500 of business expenses for AGI on her Schedule C. Her net income of 18,000 from her consulting activities is included on Schedule 1, line 3, which is then combined with other items and included on Form 1040 (page 1), line 8.

**Answer**: 10,600. Because the stock is an appreciated investment asset held for more than a year, it qualifies as capital gain property. **Answer**: 10,600. Stock is intangible property, not tangible personal property. Hence, Courtney would deduct the fair market value of the stock.

Courtney performs architectural consulting services outside of her employment with EWD and also generates rental income from a condo she owns. Will her income from the architectural services and condo rental be considered qualified business income?

**Answer**: Yes for the architectural services and possibly for the condo rental. Architectural services are not a specified service trade or business. **What if**: Assume Courtney decides to establish a sole proprietorship to perform investment advising services.
