**Example 6-1**

Suppose that Courtney purchased a parcel of land for its appreciation potential. Would her ownership in the land be considered a business or investment activity?

**Answer:** Courtney’s activity would most likely be considered an investment activity because she acquired the land for its appreciation potential and she does not plan to exercise any special effort to develop the property or to become actively involved in other real estate speculation.

**What if:** Suppose that Courtney frequently buys and sells land or develops land to sell in small parcels to those wanting to build homes. Would Courtney’s activity be considered a business or investment activity?

**Answer:** Courtney’s activity would most likely be considered a business activity because she is actively involved in generating profits from the land by developing it rather than simply holding the land for appreciation.

**Example 6-2**

Besides being employed by EWD, Courtney is also a self-employed architectural consultant (a business activity). This year her consulting activity generated 19,500 of revenue and incurred 1,500 in expense (primarily travel and transportation expenses). How does she report the revenue and deductions from the activity?

```python
business_revenue = 19500
business_expenses = 1500
net_income = business_revenue - business_expenses
print(f"{"Business Revenue":18} | {business_revenue:>6,.0f}")
print(f"{"Business Expenses":<18} | {business_expenses:>6,.0f}")
print(f"{"-"*18:<18} | {"-"*6:<6}")
print(f"{"Net Income":<18} | {net_income:>6,.0f}")
```

**Answer**: Courtney reports the 19,500 of revenue and deducts the $1,500 of business expenses for AGI on her Schedule C. Her net income of 18,000 from her consulting activities is included on Schedule 1, line 3, which is then combined with other items and included on Form 1040 (page 1), line 8.

**Example 6-3**

Courtney owns a condominium that she rents to tenants. This year she received 14,000 in rental revenue and incurred 9,000 of expenses associated with the rental, including management fees, maintenance, and depreciation. How does she report the revenue and expenses for tax purposes?

```python
rental_receipts = 14000
rental_expenses = 1500
net_rental_income = rental_receipts - rental_expenses
print(f"{"Rental Receipts":18} | {rental_receipts:>6,.0f}")
print(f"{"Rental Expenses":<18} | {rental_expenses:>6,.0f}")
print(f"{"-"*18:<18} | {"-"*6:<6}")
print(f"{"Net Rental Income":<18} | {net_rental_income:>6,.0f}")
```

**Answer:** Courtney reports the 14,000 of rental receipts and deducts the 9,000 of rental expenses for AGI on Schedule E. The net rental income of 5,000 is then reported on Schedule 1 (line 5) and combined with other items on page 1 (line 8) of her Form 1040.

**Example 6-4**

**What if**: Suppose that Courtney invested in a partnership and that this year her share of business losses from the partnership amounts to 400,000. Assume that the 400,000 loss is deductible under the basis, at-risk amount, and passive loss limitations. How much of the loss would be deductible on her tax return?

```python
single_threshold = 305000
partnership_loss = 400000
deductible = min(single_threshold, partnership_loss)
carry_forward = partnership_loss - deductible
print(f"Deductible business loss:     $ {deductible:,.0f}")
print(f"Carry forward business loss:  $  {carry_forward:,.0f}")
```

**Answer:** 305,000. Deductible net business losses are limited to 305,000 (610,000 for taxpayers married filing jointly). The 95,000 excess business loss would be carried forward to next year.

**Example 6-7**

**What if**: Courtney is self-employed and her only health insurance coverage for the entire year is by a family health plan with a 4,000 annual deductible and maximum deductible and out-of-pocket expenses of 8,000. If Courtney contributes 6,300 to an HSA to pay for medical expenses, how much of the HSA contributions could Courtney deduct as a for AGI deduction?

**Answer**: 6,300. Because Courtney is only covered by a high-deductible plan (minimum annual deductible of 3,200 for family coverage and maximum annual deductible and other out-of-pocket expenses not exceeding 16,100 for family coverage), she may fully deduct her HSA contributions up to 8,300. Hence, she may fully deduct her 6,300 of HSA contributions on Form 8889 and Schedule 1 (line 13).
