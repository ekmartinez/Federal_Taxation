
Camille Sikorski was divorced in 2018. She currently provides a home for her 15-year-old daughter Kaly.
```python
# Data Gathering
# Tax year: 2025

age = 45                                                        # N/A 
status = "HOH"                                                  # Filing status
salaries = 105000                                               # Given
alimony = 10000                                                 # Pre-2019
gross_income = salaries + alimony                               #
qualified_business_income = 0                                   # N/A
adjustments = 6000                                              # Retirement account
itemized_deduction = 15000                                      # HOH 2025 - 23,625
qbi_deduction = 0                                               # N/A
capital_gains = 0                                               # N/A
capital_gains_tax = 0                                           # N/A 
child_tax_credit = 2200                                         # 2025 qualifying child       
credit_for_other_dependent = 0                                  # N/A
withholdings = 0                                                # N/A

income_tax = OrdinaryIncomeTax(status, age, gross_income, adjustments, itemized_deduction, qbi_deduction).calculate_tax()
```

```python
print(f"Camille's taxable income is: ${income_tax["taxable_income"]:,.0f}")
```
